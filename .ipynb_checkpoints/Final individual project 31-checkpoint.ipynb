1) Data Description:
players.csv contains 7 columns/variables: experience (Beginner, Amateur, Regular, Pro, and Veteran), subscribe (TRUE or FALSE), hashed email (string), played_hours (float), name (string), gender (Male, Female, Non-Binary, Prefer not to say, Agender, Two-Spirited, or Other), and Age (natural). Each row/observation represents a player and there are 196 of them. played_hours and Age have means of 5.90 and 21.14 respectively. Some issues with the data are that there are some NAs and there are some outliers, such as people who have 0 hours played and others who have 200+ hours played. From the server sign up page, it can be seen that they collected the data by asking people for their info when they signed up and also tracking their hours played.

2) Questions: this project will address question 1: "What player characteristics and behaviours are most predictive of subscribing to a game-related newsletter, and how do these features differ between various player types?", specifically, can players' total play time and age predict whether or not they subscribe to a game-related newsletter? The data will be used to answer the question by plotting age vs. hours played and color coding the true and false points to reveal trends. From there, predictive analysis can be preformed. To wrangle this data the NA rows will be filtered out.

```r
library(tidyverse)
```

```r
players <- read_csv("players.csv")
```

```r
players_filtered <- players |>
    filter(!is.na(Age))
```

```r
players_means <- players_filtered |>
    summarize(mean_played_hours = mean(played_hours, na.rm = TRUE),
              mean_age = mean(Age, na.rm = TRUE))

players_means
```

```r
age_distribution <- ggplot(players_filtered, aes(x = Age)) +
    geom_histogram(binwidth = 2, fill = "purple", color = "black") +
    labs(title = "Distribution of Player Ages",
         x = "Age (years)",
         y = "Number of players")

age_distribution
```

```r
age_distribution <- ggplot(players_filtered, aes(x = played_hours)) +
    geom_histogram(binwidth = 2, fill = "magenta", color = "black") +
    labs(title = "Distribution of hours played",
         x = "Hours played",
         y = "Number of players")

age_distribution
```

```r
played_hrs_vs_age <- ggplot(players_filtered, aes(x = Age, y = played_hours)) +
    geom_point() +
    labs(title = "The effect of age on hours played",
         x = "Age (years)",
         y = "Hours Played")

played_hrs_vs_age
```

3) Exploratory Data Analysis and Visualization: The two distributions above show that player ages cluster around 20 and under while hours played cluster around 0 hours. The negative effects of these skewed distributions can be seen in the age vs. hours played plot, where a majority of the dots fall at 0. Based on these visualizations, it may be best to filter out the 0 hour players so that it's easier visualize any trends.

4) Methods and Plan: To address the question of "Can players' total play time and age predict whether or not they subscribe to a game-related newsletter?", the k-nearest neighbors classification method will be used. This method is appropriate for the data because the true or false answer of whether or not a player has subscribed to the newsletter is binary and can act as the positive or negative classification of a point. To preform k-nn, it must be assumed that the data is scaled correctly, i.e. hours played doesn't have a bigger impact on the prediction than age simply because it has a larger range of numbers. K-nn's weaknesses are that it is sensitive to outliers and it takes a lot of tuning to find the best k value to use. After filtering out outliers and standardizing the data, it will be split into a training set with 70% of the observations, and a testing set with 30% of the observations. Then, the training data will be split into 5 folds to preform cross-validation and find the best k value. Applying the best k value to the test set, hopefully the model's predictions are accurate.
